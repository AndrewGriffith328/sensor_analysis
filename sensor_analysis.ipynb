### Task 1: Import NumPy


```python
import numpy as np
print(np.__version__)
```

### Task2: Create the Sensor Array

```python
temps = np.array([57, 55, 54, 53, 53, 54, 56, 60,
65, 70, 74, 78, 81, 83, 84, 84,
82, 79, 74, 70, 66, 63, 61, 59])
print(temps)
print(temps.shape)
print(temps.dtype)
```

### Task 3: Convert to Celsius

```python
temps_c = (temps - 32) * 5 / 9

print(temps_c)
```

### Task 4: Summarize the Day

```python
sum = np.sum(temps)
Mean = np.mean(temps)
Min = np.min(temps)
Max = np.max(temps)
Median = np.median(temps)


print("Sum:", sum)
print("Mean:", Mean)
print("Min:", Min)
print("Max:", Max)
print("Median:", Median)
```

### Task 5: What was the warmest hour?

```python
print(f"2:00 PM temperature:{temps[14]}")
print(f"11:00 PM temperature: {temps[-1]}")
```

### Task 6: Slice out the afternoon

### Task 8: Build a 2D Sensor Table

```python
sensor_grid = np.array([[81, 83, 84, 84, 82, 79], [78, 80, 82, 83, 81, 77], [83, 85, 87, 88, 86, 82]])
print(sensor_grid)
print(sensor_grid.shape)
```

### Task 9: Read one sensor and one hour

```python
print(sensor_grid[2,2])
print(sensor_grid[1,:])
print(sensor_grid[:,0])
```

### Task 10: Mean temperature per sensor

```python
print(f"Sensor 1 mean: {np.mean(sensor_grid[0,:])} F")
print(f"Sensor 2 mean: {np.mean(sensor_grid[1,:])} F")
print(f"Sensor 3 mean: {np.mean(sensor_grid[2,:])} F")

print(sensor_grid)
print(sensor_grid[:,3])
```

### AT Task 11: Temperature range per hour

```python
hour_1 = np.max(sensor_grid[:,0]) - np.min(sensor_grid[:,0])
hour_2 = np.max(sensor_grid[:,1]) - np.min(sensor_grid[:,1])
hour_3 = np.max(sensor_grid[:,2]) - np.min(sensor_grid[:,2])
hour_4 = np.max(sensor_grid[:,3]) - np.min(sensor_grid[:,3])
hour_5 = np.max(sensor_grid[:,4]) - np.min(sensor_grid[:,4])
hour_6 = np.max(sensor_grid[:,5]) - np.min(sensor_grid[:,5])
print(hour_1)
print(hour_2)
print(hour_3)
print(hour_4)
print(hour_5)
print(hour_6)

hourly_range = [hour_1, hour_2, hour_3, hour_4, hour_5, hour_6]


x = np.array(hourly_range)
print(x)

```

### Task 12: Normalize the afternoon readings

```python
afternoon_v = np.array(afternoon)
# calculate min
afternoon_min = np.min(afternoon)
#calculate max
afternoon_max = np.max(afternoon)
#create new array using formulas above
afternoon_norm = (afternoon_v - afternoon_min) / (afternoon_max - afternoon_min)
print(afternoon_norm)
```
